```python
import pandas as pd
import matplotlib.pyplot as plt

student=pd.read_csv('student_IQdata.csv')
student
```

# left and right skewed discribution

**left skewed distribution**

```python
plt.title('student iq histogram')
plt.xlabel('iq')
plt.ylabel('no of student')
plt.hist(student['IQ_Score'],color='yellow' ,edgecolor='black')
plt.show()
```

right skewed distribution

```python
plt.title('student iq histogram')
plt.xlabel('Study hours')
plt.ylabel('no of student')
plt.hist(student['Study_Hour'],color='yellow' ,edgecolor='black')
plt.show()
```

# attributes/ parameter

bins

```python
plt.title('student iq histogram')
plt.xlabel('Study hours')
plt.ylabel('no of student')
plt.hist(student['Study_Hour'],color='green' ,edgecolor='black',bins=[1,2,3,4,5,6,7,8,9]) #1-9 porjonto se amake dekhacche
plt.show()
```
