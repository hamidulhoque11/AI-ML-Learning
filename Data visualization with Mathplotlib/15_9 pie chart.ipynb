```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('student_data.csv')
df
```

pie, labels

```python
status=df.groupby('CompletionStatus').size()
status
plt.pie(status,labels=status.index)
plt.show()
```

```python
status=df.groupby('CompletionStatus').size()
status
plt.pie(status,labels=status.index,autopct='%1.1f%%',colors=['rosybrown','green','red'],shadow=True,explode = (0.1, 0, 0) )
plt.show()
```
