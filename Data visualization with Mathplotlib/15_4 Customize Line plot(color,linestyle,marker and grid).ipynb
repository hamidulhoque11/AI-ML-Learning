```python

import pandas as pd
import matplotlib.pyplot as plt
data={
   'Year' : [2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],
    'Programming' : [589,862,1085,1150,1439,1715,1785,1995,2236,2442],
    'Digital Marketing' : [735,963,1157,1404,1497,1579,1948,2073,2310,2338]
}
df=pd.DataFrame(data)
df
```

```python
plt.xlabel('Year')
plt.ylabel('programming and digital marketing')
plt.title('relation between prog and digi. marketing')
plt.plot(df['Year'],df['Programming'],label='programming',color='red',linewidth='1',linestyle='dotted',marker='o',markersize=7)
plt.plot(df['Year'],df['Digital Marketing'],label='digital marketing',color='black',linewidth=1,linestyle='dotted',marker='o')
plt.legend()
plt.grid()
plt.show()
df
```
