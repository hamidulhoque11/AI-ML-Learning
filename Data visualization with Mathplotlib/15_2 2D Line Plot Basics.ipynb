x and y axis add Name

```python
import matplotlib.pyplot as plt
Study_hours=[2,3,4,1,5,7,3]
Study_days=[1,2,3,4,5,6,7]
plt.xlabel('Study Days')
plt.ylabel('Study Hours')
plt.title('Study Time over a week')
plt.plot(Study_days,Study_hours)
```

```python
# ekhane 2 jon student er data nilam
import matplotlib.pyplot as plt
Study_hours1=[2,3,4,1,5,7,3]
Study_hours2=[1,4,1,2,5,7,2]
Study_days=[1,2,3,4,5,6,7]
plt.xlabel('Study Days')
plt.ylabel('Study Time')
plt.title('Two student Study time over week')
plt.plot(Study_days,Study_hours1)
plt.plot(Study_days,Study_hours2)
```
