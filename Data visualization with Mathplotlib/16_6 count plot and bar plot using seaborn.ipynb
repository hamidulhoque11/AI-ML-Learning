## countplot

```python
import pandas as pd
import numpy as np
import seaborn as sns
data=pd.read_csv('sns_data.csv')
data.head(3)
```

```python
sns.countplot(data=data,x='gender')
```

```python
sns.countplot(data=data,x='subject')
```

```python
sns.countplot(data=data,x='subject',hue='gender')
```

## Barpolot

```python
student=pd.read_csv('student_data.csv')
student.head(3)
```

```python
sns.barplot(data=student,x='Instructor',y='Python Marks')
```

```python
sns.barplot(data=student,x='Instructor',y='Python Marks',errorbar=None)
```

```python
sns.barplot(data=student,x='Instructor',y='Python Marks',errorbar=None,estimator=np.median)
```

```python
sns.barplot(data=student,x='Instructor',y='Python Marks',errorbar=None,estimator=np.max)
```
