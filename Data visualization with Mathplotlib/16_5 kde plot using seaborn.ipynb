```python
import seaborn as sns
tips_data=sns.load_dataset('Tips')
tips_data
```

```python
sns.displot(kind='hist' , data=tips_data,x='total_bill')
```

```python
sns.displot(kind='hist' , data=tips_data,x='total_bill',bins=20)
```

```python
sns.displot(kind='hist' , data=tips_data,x='tip',)
```

## kde plot using seaborn

```python
sns.kdeplot(data=tips_data, x='total_bill')
```

```python
sns.kdeplot(data=tips_data, x='total_bill',hue='sex')
```

```python
sns.kdeplot(data=tips_data, x='total_bill',hue='sex',fill=True)
```
