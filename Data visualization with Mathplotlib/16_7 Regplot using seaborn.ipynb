## regplot

```python
#  ases level
import seaborn as sns
import pandas as pd
data=pd.read_csv('sns_data.csv')
data.head(3)
```

```python
sns.regplot(data=data,x='study_hours', y= 'test_score')
```

```python
sns.regplot(data=data,x='study_hours', y= 'test_score',hue='gender')# we cont use hue
```

## fegure  level : lmplot

```python
sns.lmplot(data=data,x='study_hours',y= 'test_score')
```

```python
sns.lmplot(data=data,x='study_hours',y= 'test_score',hue='gender')
```
