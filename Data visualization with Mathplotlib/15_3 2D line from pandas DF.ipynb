# 2D Line from DataFrame

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
   'Year' : [2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],
    'Programming' : [589,862,1085,1150,1439,1715,1785,1995,2236,2442],
    'Digital Marketing' : [735,963,1157,1404,1497,1579,1948,2073,2310,2338]
}
df=pd.DataFrame(data)
df
```

```python
plt.xlabel("Year")
plt.ylabel('Programming')
plt.title("Student Enrollment over 8 year")
plt.plot(df['Year'],df['Programming'])
```

```python
plt.xlabel("Year")
plt.ylabel('Digital Marketing')
plt.title("Student Enrollment over 8 year")
plt.plot(df['Year'],df['Digital Marketing'])
```

```python
plt.xlabel("Year")
plt.ylabel('Digital Marketing')
plt.title("Student Enrollment over 8 year")
plt.plot(df['Year'],df['Programming'])
plt.plot(df['Year'],df['Digital Marketing'])
```

# lagent()

lagent use korchi ekhane kontar plot konta ba straight line konta boja jacche na tai 

```python
plt.xlabel("Year")
plt.ylabel('Digital Marketing')
plt.title("Student Enrollment over 8 year")
plt.plot(df['Year'],df['Programming'],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')
plt.legend()
```
