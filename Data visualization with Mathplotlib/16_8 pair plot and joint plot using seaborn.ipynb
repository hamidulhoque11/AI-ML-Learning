```python
#  ases level
import seaborn as sns
import pandas as pd
data=pd.read_csv('sns_data.csv')
data.head(3)
```

```python
student_marks=data[['test_score','study_hours','attendance_rate','gender']]
student_marks.head(3)
```

```python
sns.pairplot(data=student_marks)
```

## histogram


```python
sns.pairplot(kind='hist',data=student_marks)
```

```python
sns.jointplot(data=student_marks,x='study_hours',y='test_score')
```

```python
sns.jointplot(data=student_marks,x='study_hours',y='test_score',kind='kde')
```

```python
sns.jointplot(data=student_marks,x='study_hours',y='test_score',kind='kde',hue='gender')
```
