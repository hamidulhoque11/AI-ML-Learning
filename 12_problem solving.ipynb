```python
num=input().split()

#print(num)
x=int (num[0])
y=int (num[1])
z=int (num[2])

max=x
min=x

if y>max:
    max=y
if z>max:
    max=z

if y<min:
    min=y
if z<min:
    min=z

print(min,max)
```

```python
##Even odd possitive and negetice
n=int(input()) #no of input
value=input()
num=value.split()
positive=0;negative=0
even=0;odd=0
for x in range(n):
    x= int (num[x])
    if x>0:
        positive +=1
    elif x<0:
        negative+=1
    if x%2==0:
        even+=1
    else:
        odd+=1
print("Even:",even)
print("Odd:",odd)
print("Positive:",positive)
print("Negative:",negative)
```
