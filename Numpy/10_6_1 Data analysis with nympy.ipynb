zeros

```python
import numpy as np

arr=np.zeros((3,3))
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr3=np.array([
    [10,20,30],
     [30,20,1]
])
print(arr3)
```

zeros like

```python
arr=np.zeros_like(arr3)
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
