1D matrix

```python
#In-place-soring
import numpy
x=numpy.array([10,100,8,16,1,10])
print(x)
x.sort() #inplace soring
print(x)

print(x) #ekhane problem ta holo x ke amra ar use korte parbona
```

```python
#copy-soring
import numpy
x=numpy.array([10,100,8,16,1,10])
print(x)
copy_soring=numpy.sort(x)
print(copy_soring)

print(x) #porobortite amra x er valuetake egain use korte parbo
```

2d matrix soring

```python
arr=numpy.array([[10,2,5],[8,4,9]])
print(arr)
print(arr.shape)
print(arr.ndim)
```

Horizontal soring

```python
print(arr)
```

```python
horizontal_soring=numpy.sort(arr,axis=1)
print(horizontal_soring)
```

```python
vertical_soring=numpy.sort(horizontal_soring,axis=0)
print(vertical_soring)
```

```python
whole_sort=numpy.sort(arr) #ekhane axis value na dile auto horizontal akare sorting hoye jay
print(whole_sort)
```
