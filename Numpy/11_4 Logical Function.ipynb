```python
import numpy
a=numpy.array([10,20,30,40])
b=numpy.array([100,5,30,35])

greater_then=a>b
print(greater_then)

less_then=a<b
print(less_then)

equal_to=a==b
print(equal_to)
```

all function

```python
print(a)
print(b)
print(equal_to)
all_equal=numpy.all(equal_to) #sob golo true hote hobe
print(all_equal)
```

```python
print(a)
b=numpy.array([10,20,30,40])
print(a==b)
all_equal=numpy.all(a==b)
print(all_equal)
```

any function

```python
print(a)
b=numpy.array([10,20,30,40])
print(a==b)
all_equal=numpy.any(a!=b) #jekono ekta true holei true hobe 
print(all_equal)
```

```python
print(a)

b=numpy.array([10,20,30,49])
print(b)
print(a==b)
all_equal=numpy.any(a!=b) 
print(all_equal)
```
