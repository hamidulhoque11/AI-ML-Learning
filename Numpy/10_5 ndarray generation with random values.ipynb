rand

```python
import numpy as np
arr=np.random.rand(2,3)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

randint

```python
import numpy as np
arr=np.random.randint(0,3,(2,3))#0 and 3 is a starting and ending and 2,3 is shape
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

we also can play with it

```python
arr=np.random.randint(0,6,1)
arr=arr.astype(np.int8)
value=0
for x in arr:
    value=x
print(value)
print(type(value))
guess_game=int(input("Enter a value 0-5 :"))
if guess_game==value:
    print("Win")
    print("your random value is:",value)
else:
    print("you are lose")
    print("your random value is:",value)
```

uniform

```python
#using unifrom for float value
import numpy as np
arr=np.random.uniform(0,3,(3,2,3))#we can also change diamansion
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
