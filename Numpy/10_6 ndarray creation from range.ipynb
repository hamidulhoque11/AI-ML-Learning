Range:

```python
import numpy as np
arr=np.arange(0,10,1)
print(arr)
```

```python
import numpy as np
arr=np.arange(2,10,2)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

matrix and reshape() function

```python
import numpy as np
arr=np.arange(1,10,1).reshape(3,3)
print(arr)
```

```python
import numpy as np
arr=np.arange(1,10,1)
mat=arr.reshape(3,3) #value te rekheo funtion ke call korete pari
print(mat)
print(type(mat))
print(mat.dtype)
print(mat.shape)
print(mat.ndim)
print(mat.size)
```

linear / linspace() function

```python
#linspace(staring , ending, pick point with range)

import numpy as np
arr=np.linspace(1,10,10)#1-10 er jonne 10 ta point
print(arr)
```

```python
import numpy as np
arr=np.linspace(1,10,11) #ei range er oddhe 11 ta point se kore diche
print(arr)
```

```python
import numpy as np
arr=np.linspace(1,10,15)#same range er moddhe 15 tapoint kore dicche
print(arr)
```
