1D array indexing


```python
import numpy as np
#1D array
arr1=np.arange(1,6)
print(arr1)
print(arr1[2])
arr1[2]=100
print(arr1)
arr1[0]=50
print(arr1)
```

slicing:

```python
print(arr1)
arr=arr1[0:3]
print(arr)
```

matrinx indexing

```python
import numpy as np
#2D array
arr2=np.arange(1,10).reshape(3,3)
print(arr2)
```

```python
print(arr2[0][2]) # access korte parchi
arr2[1][1]=50 #change oo korte parchi
print(arr2)
print()
arr2[2][1]=150
print(arr2)
```

slicing

```python
#row 0
row_0=arr2[0:1,]
print(row_0)
```

```python
#row 1
row_1=arr2[1:2,]
print(row_1)
```

```python
#row 2
row_2=arr2[2:,]
print(row_2)
```
