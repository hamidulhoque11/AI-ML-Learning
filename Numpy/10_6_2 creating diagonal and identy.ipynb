diagonal (primary and secondary)

```python
import numpy as np
img=np.array([
    [1,2,3],
    [4,5,6],
    [8,9,10]
    ])
print(img)
print(img.ndim)
print(img.shape)
```

```python
diagonal=np.diag([1,5,10])
print(diagonal)
```

```python
diagonal=np.diag([1,5,10,7])
print(diagonal)
```

```python
diagonal=np.diag([1,5,10,7,0])
print(diagonal)
```
