list to ndarray

```python
lst=[10,20,30,40]
print(lst)
```

```python
import numpy as np
lst=[10,20,30,40]
arr=np.array(lst)
print(arr)
print(arr.dtype)
print(type(arr))
print(arr.shape)
print(arr.size)
print(arr.ndim)
```

```python
import numpy as np
lst=[10,20,30,40]
arr=np.array(lst,dtype=np.float64)
print(arr)
print(arr.dtype) #chaile data type taw change korte pari
print(type(arr))
print(arr.shape)
print(arr.size)
print(arr.ndim)
```

```python
import numpy as np
mixed_lst=[10,True,30,40.10,"Hamidul"]
arr=np.array(mixed_lst)
print(arr)
print(arr.dtype)  #data type string
print(type(arr))
print(arr.shape)
print(arr.size)
print(arr.ndim)
```

```python
import numpy as np
mixed_lst=[10,True,30,40.10,"Hamidul"]
arr=np.array(mixed_lst)
arr=arr.astype(np.float64) #chailei float ee convert korte parnona karon eta string data type
print(arr)
print(arr.dtype)  #data type string
print(type(arr))
print(arr.shape)
print(arr.size)
print(arr.ndim)
```
