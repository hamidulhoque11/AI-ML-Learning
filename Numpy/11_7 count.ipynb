```python
import numpy as np
a=np.random.randint(1,100,size=(100,))
print(a)
```

```python
soring_a=np.sort(a)
print(soring_a)
```

```python
count=np.count_nonzero(a==85) #a er moddhe koyta 85 ache ta dekhabe
print(count)
```

```python
count=np.count_nonzero(a==1) #a er moddhe koyta 85 ache ta dekhabe
print(count)
```

```python
count=np.count_nonzero(a>85) # a er moddhe 85 upore koyta value asche
print(count)
```

```python
count=np.count_nonzero(a<=85) #a er moddhe 85 soman ba niche koyta value ache
print(count)
```
