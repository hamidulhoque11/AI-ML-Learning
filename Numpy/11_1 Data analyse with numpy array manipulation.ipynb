Concatenate

```python
#col
a=np.random.randint(1,10,size=(1,3))
b=np.random.randint(1,5,size=(1,3))
print(a)
print()
print(b)
print()
con_A_and_B=np.concatenate((a,b),axis=1)
print(con_A_and_B)
```

```python
#ow
a=np.random.randint(1,10,size=(3,1))
b=np.random.randint(1,5,size=(3,1))
print(a)
print()
print(b)
print()
con_A_and_B=np.concatenate((a,b),axis=0)
print(con_A_and_B)
```

```python
a=np.random.randint(1,10,size=(2,3))
b=np.random.randint(1,5,size=(1,3))
print(a)
print()
print(b)
```
