1d array

```python
import numpy as np
a=np.array([2,5,10,8,16,100])
print(a)
```

where() function

```python
index=np.where(a>5)
print(index)
```

```python
print(a)
index=np.where(a==10)
print(index)
```

```python
change = np.where(a>10,0,a)
print(change)
```

```python
change = np.where(a>10,a,7)
print(change)
```

```python
print(a)
index=np.where(a==5)
print(index)
```

find maximum value's index

```python
print(a)
max_index=np.argmax(a)
print("Max value:",max_index)
```

```python
print(a)
min_value_index=np.argmin(a)
print("min value: ",min_value_index)
```

matrix search

```python
#2d array
matrix=np.array([[10,3,5],[8,4,9]])
print(matrix)

find=np.where(matrix==4)
print(find) # x and y axis ee show korbe 
```

```python
print(matrix)
find=np.where(matrix>4)
print(find)
```

```python
print(matrix)
print()
find=np.where(matrix>4,matrix,0)
print(find)
```

```python
print(matrix)
print()
find=np.where(matrix>4,0,matrix)
print(find)
```
