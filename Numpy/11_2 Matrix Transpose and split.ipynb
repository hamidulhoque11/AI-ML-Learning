T Function

```python
import numpy as np
arr=np.arange(1,10).reshape(3,3)
print(arr)
```

```python
transpose=arr.T
print(transpose)
```

Array split

```python
print(arr)
```

```python
splited_array=np.array_split(arr,3)
print(splited_array)
```

```python
splited_array=np.split(arr,3)
print(splited_array)
```
