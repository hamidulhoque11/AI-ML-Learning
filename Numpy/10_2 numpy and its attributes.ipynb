diclaration

```python
import numpy as np
#1D array
arr1=np.array([1,2,3])
print(arr1*5)
print(type(arr1))
print(arr1.ndim)
```

```python
#2d array
arr2=np.array([
    [1,2,3],
    [4,5,6]
])
print(arr2)
print("diamantion : ",arr2.ndim)
```

```python
#3d array
arr3=np.array([
    [
       [1,2,3],
       [1,2,3]
    ]
])
print(arr3)
print("diamention: ",arr3.ndim)
print(arr3.shape)
```

```python
#3d array
arr4=np.array(
    [
       [
           [1,2,3],
            [1,2,3]
       ],
       [
          [1,2,3],
          [1,2,3] 
       ]
    ]
)
print("diamantion : ",arr4.ndim)
print(arr4)
print(arr4.shape)
```

array attributes

```python
#dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

#shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
print(arr4.shape)

#data type:
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size:
print(arr1.size)
print(arr2.size)
print(arr3.size)
print(arr4.size)
```

```python
#3d array

arr3 = np.array([
    #1st floor
    [ [1,2,3], # row1
      [4,5,6]  ], #row2

    #2nd floor

     [[1,2,3], # row1
      [4,5,6]  ], #row2


    #3rd floor
      [ [1,2,3], # row1
      [4,5,6]  ] #row2


])
print(arr3.size)
print(arr3)
print("dimension: ",arr3.ndim)
print(arr3.shape)
```
