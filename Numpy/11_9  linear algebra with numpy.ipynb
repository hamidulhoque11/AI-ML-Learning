```python
import numpy as np
a=np.array([
    [1,2,3],
    [4,5,6]
])
b=np.array([
    [7,8],
    [9,10],
    [11,12]
])
```

```python
c=np.array([
    [1,2,3],# 1+5+3 =9
    [4,5,6],
    [1,2,3]
])
squate_matrix=np.trace(c) 
print(squate_matrix)
```

Determinent

```python
c=np.array([
    [1,2,3],# 1(5*3-6*2)-2(4*3-6*1)+3(4*2-5*1)
    [4,5,6],
    [1,2,3]
])

determinent_A=np.linalg.det(c)
print(determinent_A)
```

rank


```python
rank_sq=np.linalg.matrix_rank(c)
print(rank_sq)
```
