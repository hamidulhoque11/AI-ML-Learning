genform(location,delimiter=',',and more)

```python
#data load
import numpy
data=numpy.genfromtxt('11_8 mraks data.csv',delimiter=',',skip_header=1)
print(data)
```

slicing math column

```python
math_marks=data[::,1] # data theke math col ta slicing kore niye aschi
print(math_marks)
```

find max value

```python
max_math_marks=numpy.max(math_marks)
print(max_math_marks)
```

find minimum marks

```python
min_math_marks=numpy.min(math_marks)
print(min_math_marks)
```

avarage / mean

```python
mean_math_marks=numpy.mean(math_marks)
print(mean_math_marks)
```

median

```python
median_math_marks=numpy.median(math_marks)
print(median_math_marks)
```

sort

```python
sort_math_marks=numpy.sort(math_marks)
print(sort_math_marks)
```

Standard Diviation

```python
std_math_marks=numpy.std(math_marks)
print(std_math_marks)
```
