selecting for an array

```python
import numpy as np
arr =np.array([1,2,3.40,"hamidul"],dtype=np.float64) #string ke float ee rakhte parbona tai vlaue error
print(arr)
print(arr.dtype)#string
```

```python
import numpy as np
arr =np.array([1,2,3.40,],dtype=np.int8) #amra -127 theke 127 porjonto value store kore rkahte parbo
print(arr)
print(arr.dtype)#string
```

```python
import numpy as np
arr =np.array([1,2,300,],dtype=np.int8) # OverflowError   karon int8 -127 theke +127 porjonto value store kore rakha jay,ekhane 300 > 127 tai overflow error asche
print(arr)
print(arr.dtype)#string
```
