Arithmetic operator

```python
x=[10,20,30]
y=[1,2,3]
print(x,y)
add=x+y
print("add",add)
```

```python
import numpy as np
x=np.array([10,20,30,40])
y=np.array([1,2,3,4])
print(x)
print(y)
add=x+y
print(add)
```

```python
#addition
add=x+y
print("Addition:",add)

#substraction
sub=x-y
print("Substraction",sub)

#Multiplication
mul=x*y
print("MUltipication",mul)

#Divition
div=x/y
print("Divition",div)

#reminder
rem=x%y
print("reminder",rem)

```

cos er value ke calculator er moto pete hole

```python
#deg to radian
print(x)
radian=np.deg2rad(x)
print(radian)
cos=np.cos(radian)
print(cos)
```

```python
#radian to deg and deg to radian
print(x)

#degree to radian
deg_to_radian=np.deg2rad(x)
print("degree to radian: ",deg_to_radian)

#radian to degree
radian_to_degree = np.rad2deg(deg_to_radian)
print("Radian to degree",radian_to_degree)
```
