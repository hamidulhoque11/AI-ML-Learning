Problem 1: Smart Light
A smart light turns ON if the room is bright enough.
You are given two integers:
brightness
threshold
If brightness >= threshold, print ON. Otherwise, print OFF.
Input Format
Two integers: brightness and threshold
Output Format
ON
or
OFF
Sample Input
75 60
Sample Output
ON


```python
n=input().split()
brightness=int(n[0])
threshold=int (n[1])
if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```

Problem 2: Even or Odd
You are given an integer N.
Determine whether the number is Even or Odd.
Input Format
One integer N
Output Format
Even
or
Odd
Sample Input
17
Sample Output
Odd


```python
x=int(input())
if x%2==0:
    print("Even")
else:
    print("Odd")
```

Problem 3: Print Numbers
You are given an integer N.
Print all numbers from 1 to N, one number per line.
Input Format
One integer N
Sample Input
5
Sample Output
1
2
3
4
5


```python
n=int(input())
for x in range(1,n+1):
    print(x)
```
