**1. Add New Column and Save File**

- You have a DataFrame with column Price.

**3. Handle Missing Values**

Columns: Age, Salary

 Tasks:

- Count missing values


- Fill missing Age with mean age


- Drop rows where Salary is missing


```python
import pandas as pd
df=pd.read_csv('15_Practice.csv')
count_missing_age=df['Age'].isnull().sum()
print("Missing Age:",count_missing_age)
count_missing_Salary=df['Salary'].isnull().sum()
print("Missing Salary:",count_missing_Salary)
new=df['Age'].fillna(df['Age'].mean())
df['Age']=new
df
```

```python
df.dropna(subset='Salary')
```

# problem 4

4. Apply Statistical Functions

Columns: Marks1, Marks2, Marks3

 Tasks:

- Calculate mean, max for each subject

- Compute total marks per student


- Find standard deviation of total marks


```python
import pandas as pd
df=pd.read_csv('15_Practice.csv')
#mean
print("Mean (marks 1):",df['Marks1'].mean().__round__(2))
print("Mean (marks 2):",df['Marks2'].mean().__round__(2))
print("Mean (marks 3):",df['Marks3'].mean().__round__(2))
print()
#max
print("Max (marks 1):",df['Marks1'].max())
print("Max (marks 2):",df['Marks2'].max())
print("Max (marks 3):",df['Marks3'].max())
print()
#Celculate the total marks for per students
#row wise
total_marks=df[['Marks1','Marks2','Marks3']].sum()
print()
#column wise
total_marks=df[['Marks1','Marks2','Marks3']].sum(axis=1)
print(total_marks)
#standard deviation
print()
print("Standard deviation : ",total_marks.std())
```

# problem 5

**5. Histogram, Bar, Scatter, Pie**

Create:

- Histogram of Age

- Scatter plot (Age vs Salary)

- Scatter plot ( Price vs Discounted Price) 



```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('15_discounted_products.csv')
df
```

```python
#histogram
plt.hist(df['Age'])
plt.xlabel('Age')
plt.ylabel('No of Frequencey')
plt.title('Histogram')
plt.show()
```

```python
#Scatter plt
plt.scatter(df['Age'],df['Salary'],color='red')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.show()
```

```python
plt.scatter(df['Price'],df['Discounted_Price '])
plt.xlabel('Price')
plt.ylabel('Discount Price')
plt.title("Prive vs discount price")
```
