prblem 1. Create a text file named emails.txt containing multiple email addresses, some in uppercase and some repeated.
Write a Python program to clean the data by converting all emails to lowercase, removing duplicates, sorting them alphabetically, and saving the cleaned list to a new file named unique_emails.txt.


```python
value=[]
with open("D:/play ground/AI-ML-Learning/Problem solving/07_emails.txt","r")as file:
    data=file.readlines()
    line=list(map(str.strip,data)) #line by line index ee nilam
    lower=line=list(map(str.lower,line))# list ee thaka string golo ke lower case korlab 
    single=set(lower) #conver to set for remove duplicate value
    single_list=list(single) # single value derke again list ee niye aslam
    value=single_list
with open ("D:/play ground/AI-ML-Learning/Problem solving/07_unique_emails.txt","a") as file:
    for x in value:
        file.write(x+'\n')
    
```

prblem 2. 
Create a text file named names.txt containing a list of names.
Some names may have extra spaces, inconsistent capitalization (like “aLiCe” or “ bob ”), or appear more than once.
Write a Python program to clean the data by:
1.  Removing extra spaces.
2.  Converting all names so that only the first letter is capitalized.
3.  Removing duplicate names. 
4. Sorting the names alphabetically.
Finally, save the cleaned list into a new file named clean_names.txt.


```python

with open("D:/play ground/AI-ML-Learning/Problem solving/07_names.txt","r")as file:
    data=file.readlines()
    
    #remove space
    remove_space= list(map(str.split,data))
    # add single space
    remove_space=list(map(lambda x:" ".join(x),remove_space))#joint function er maddhome amra extra space golo ke remove kore felchi
    
    #lower case
    lower=[x.lower() for x in remove_space]
    
    #Converting all first letter is capitalized
    capitalize=list(map(lambda x:x.capitalize(),lower))
    
    #removie duclicate name
    remove_duplicate=set(capitalize)
    remove_duplicate=list(remove_duplicate)
    print(remove_duplicate)
    #sorting
    sort_data=[x for x in sorted(remove_duplicate)]
    
with open("D:/play ground/AI-ML-Learning/Problem solving/07_clean_names.txt","a")as file:
    for x in sort_data:
        file.write(x+'\n')
    
    
```
