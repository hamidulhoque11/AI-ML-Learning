E. Clean Images, Then Features (Medium)

```python
n=int(input())
num=input().split()
if n==len(num):
    num=[int(x) for x in num]
    even=list(filter(lambda x:x%2==0,num))
    square=list(map(lambda x:x**2,even))
    print(square)
else:
    print("n value and length of sum must be same")
```
