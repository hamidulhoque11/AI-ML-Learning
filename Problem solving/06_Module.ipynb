D. The Model Registry (Medium)

```python
def model_info(name, layers, lr =0.01):
    return f"Model: {name}, Layers: {layers}, LR: {lr}"
n=int(input())
for x in range(n):
    value=input().split()
    if len(value) > 2:
        print(model_info(value[0],value[1],value[2]))
    else:
        print(model_info(value[0],value[1]))

```

E. Clean Images, Then Features (Medium)

```python
n=int(input())
num=input().split()
if n==len(num):
    num=[int(x) for x in num]
    even=list(filter(lambda x:x%2==0,num))
    square=list(map(lambda x:x**2,even))
    print(square)
else:
    print("n value and length of sum must be same")
```
