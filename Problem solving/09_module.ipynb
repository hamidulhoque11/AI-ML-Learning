1.  Make three classes:

Phone → has model and battery (default 100).


SmartPhone → inherits from Phone and adds operating system.


GamingPhone  → inherits from SmartPhone and adds cooling_system.
 Add a method start_game(name) that prints
   "Playing <name> on <model>".


```python
class phone:
    def battery(self,battery=100):
        self.battery=battery
        print(f"{self.battery}%")
class SmartPhone(phone):
    def operating_system():
        print("ios 27")
class GamingPhone(SmartPhone):
    def cooling_system(self,phone_temperature):
        if phone_temperature >30 : 
            print("colling fan is On")
        else:
            print("colling fan is off")
    def start_game(self,name):
        if name==name:
            print("playing on")
asus=GamingPhone()
asus.start_game("pubg")

```

2. Make a class MobilePhone with:

model (public)


__battery (private, default 100)
 Add methods:


use() → reduce battery


charge() → increase battery (max 100)


get_battery() → show battery


Show that you can’t access __battery directly.


```python
try :
    class MobilePhone:
        def __init__(self, name, battery=100):
            self.name = name
            self.__battery = battery
        def setter(self,battery):
            self.__battery=battery
        def getter(self):
            return self.__battery
        
        def use(self, reduce_battery):
            self.__battery -= reduce_battery
            if self.__battery < 0:
                self.__battery = 0
            else: 
                print(f"{self.__battery}% charge")

        def charging(self, increase_battery):
            self.__battery += increase_battery
            if self.__battery > 100:
                self.__battery = 100
            print(f"{self.__battery}% charge")
    iphon=MobilePhone('Iphone 17')
    print(iphon.__battery)
    
    
except AttributeError :
    print("You cant use battery directly")

iphon.use(50)
iphon.charging(10)
print(iphon.getter())
```
