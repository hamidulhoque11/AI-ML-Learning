Problem 2
Take an integer N. Print Even if N is even, otherwise print Odd. Use the modulus operator %. Sample Input: 17 Sample Output: Odd



```python
n=int(input())
if n%2==0:
    print("Even")
else:
    print("Odd")
```

Problem 3
Take an integer marks. Print the grade using this rule, and put the highest check first: A+ if marks >= 80, A if marks >= 70, Pass if marks >= 40, otherwise Fail. Sample Input: 85 Sample Output: A+ Sample Input 2: 40 Sample Output 2: Pass



```python
marks=int(input())
if marks >=80:
    print("A+")
elif marks >=70:
    print("A")
elif marks>=40:
    print("Pass")
```

Problem 4
Take an integer N. Use a while loop to print numbers from 1 to N, each on a new line. Do not forget to increase the counter. Sample Input: 5 Sample Output: 1 2 3 4 5



```python
n=int(input())
i=1
while i<=n:
    print(i,end=" ")
    i+=1
```

Problem 5
Take an integer N. Use a for loop and range to print all odd numbers from 1 to N. Sample Input: 10 Sample Output: 1 3 5 7 9



```python
n=int(input())
for x in range(1,n):
    print(x,end=" ")
```

Problem 6
Take an integer N. Print all numbers from 1 to N, but skip multiples of 3 using continue. Sample Input: 10 Sample Output: 1 2 4 5 7 8 10



```python
n=int(input())
for x in range(1,n+1):
    i=int(x)
    if i%3==0:
        continue
    else:
        print(i,end=" ")
```

Problem 7
Keep taking integers as input. Add them to a total. Stop when a negative number appears (break). Do not add that negative number. Then print the total. Sample Input: 4 8 15 -3 23 Sample Output: 27



```python
n=input().split()
sum=0
for x in range(len(n)):
    i=int(n[x])
    if i<0:
        break
    else:
        sum+=i
print(sum)
```

Problem 9
Take a string as input. Count how many vowels it has (a, e, i, o, u in both lowercase and uppercase) using a loop. Print only the count. Sample Input: Education Sample Output: 5



```python
n=input()
count=0
for x in range(len(n)):
    value=n[x].lower()
    if value=='a' or value=='e' or value=='i' or value=='o' or value=='u':
        count+=1
    else:
        continue
print(count)
```

Problem 10
You are given a list of marks: [90, 40, 75, 88, 35]. For each mark, print the mark and Pass if it is at least 40, otherwise Fail. Then print how many students passed and the average of all marks. Sample Input: No input Sample Output: 90 Pass 40 Pass 75 Pass 88 Pass 35 Fail Passed: 4 Average: 65.6



```python
marks=[90, 40, 75, 88, 35]
for x in range(len(marks)):
    i=int(marks[x])
    if i>=40:
        print(f"{i} Pass")
    else:
        print(f"{i} Fail")
```
