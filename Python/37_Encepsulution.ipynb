```python
class Mobile:
    def __init__(self,name,model,imei):
        self.__name=name
        self.__model=model
        self.__imei=imei #private
    def charge(self):
        print("phone is charging")
    def name_getter(self):
        return self.__name

    def model_getter(self):
        return self.__model

    def imei_getter(self): #private attribute ta ke access korar jonne moloto gatter use kora hoyeche
        return self.__imei
    def name_setter(self,name):
        self.__name=name
```

```python
Iphone=Mobile("iphone",13,15524)
print(Iphone.model_getter())
Iphone.name_setter("hamid's iphone")
print(Iphone.name_getter()) 

```
