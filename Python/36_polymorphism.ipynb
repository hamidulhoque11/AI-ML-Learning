```python
class camera:
    def __init__(self,name):
        self.name=name

    def capture (self):
        print("a photo is captured")
```

```python
class smartphone(camera):
    def __init__(self,name,resolution):
        super().__init__(name)      
        self.resolution=resolution
    def capture (self):
            print("a photo is captured by smartphone")
class dslr(camera):
    def __init__(self,name,resolution): 
        super().__init__(name)
        self.resolution=resolution
    def capture (self):
            print("a photo is captured by camera")
class drone(camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution=resolution
    def capture (self):
            print("a photo is captured by drone")
PHONE = smartphone("Iphone",30)
DSLR =dslr("DSLR",100)
DRONE = drone("Drone",30)
PHONE.capture()
DSLR.capture()
DRONE.capture()
```
