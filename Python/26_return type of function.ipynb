```python
def tup(*args):
    return args
print(tup(10,20))
print(type(tup( 10,20)))
#unzip
x,y=tup(10,20)
print(x,y)
```

```python
def dec(**kwargs):
    return kwargs
print(dec(a=10,b=20))
print(type(dec(a=10,b=20)))
```

```python
def lst(a,b):
    return [a,b]
print(lst(10,20))
print(type(lst(10,20)))
```

```python
def show_numbers(numbers):
    return numbers

lst=show_numbers([10, 20, 30, 40, 50])
print(lst)
print(type(lst))
```
