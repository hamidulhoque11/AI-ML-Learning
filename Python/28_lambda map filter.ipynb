Lambda Function

```python
#lambda parameter : expression
print((lambda a,b:a**2 +2*a*b+b**2)(2,2))
```

```python
a=lambda a,b : a**2 +2*a*b+b**2
print(a(2,2))
```

Map Function

```python
def cel(number):
    return number **2
lst=list(range(1,6))
result=list(map(cel,lst))
print(result)
```

```python
x,y=map(int, input().split())
print(x,y)
```

```python
s = ['1', '2', '3', '4','hamidul',"hamidul"]
print(s)
```

```python
s3=['1', '2', '3', '4']
s2=map(int,s3)
s2=list(s2)
print(s2)
```

```python
def myfunc(a, b):
  return a + b

x = map(myfunc, ('apple', 'banana', 'cherry'), ('orange', 'lemon', 'pineapple'))

print(list(x))
```

```python
def double(val):
    return val * 2

a = [1, 2, 3, 4]    
res = list(map(double, a))
print(res)
```

Filter Function

```python
num=[10,1,2,6,5,4]
print(list(filter(lambda x: x%2==0,num)))
print(sorted(list(filter(lambda x: x%2==0,num))))
```

```python
ages = [5, 12, 17, 18, 24, 32]
def cel(x):
    if x<18:
        return False
    else:
        return True
adult=filter(cel,ages)
for x in adult:
    print(x,end=" ")

```
