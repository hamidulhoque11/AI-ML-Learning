```python
#parants class
class phone:
    catagory="Electronics"
    #instance
    #constructor
    def __init__(self,model,battery,camera,battery_persentage=100):
        self.model=model
        self.battery=battery
        self.camera=camera
        self.battery_persentage=battery_persentage
    #method
    def charge(self,hour):
        self.battery_persentage+=hour
    def capture(self,photo):
        if(self.battery_persentage)<=0:
            print("No charge")
        else:
            self.battery_persentage-=photo
            print(f"photo capture in {self.model}")
    
#object
apple=phone("iphone 17",3000,"40MP")
blueberry=phone("b-17",4000,"30MP")
motorola=phone("M-17",3500,"35MP")
sumsung=phone("A17",5000,"30MP")
#print method
print(motorola.battery_persentage)
apple.capture(10)
apple.charge(5)
print(apple.battery_persentage)

```
