positional arguments

```python
def add(x,y):# 10,20 mapping to x,y
    print(x,y)
print(type(add(10,20)))
```

```python
def add(x,y,z):
    print(x,y,z)
add(10,20,5)
```

variable size arguments

```python
def add_arg(*args):
    print(args)
add_arg(10,20,55,7,4,1)
```

```python
def add_arg(*args):
    for x in args:
        print(x,end=" ")
add_arg(10,20,55,7,4,1)
```

keyword arguments

```python
def add_kwargs(**kwargs):
    print(kwargs)
add_kwargs(a=10,b=10,c=5,z=100,book="atomic habit")
```

```python
def add_kwargs(**kwargs):
    for k,v in kwargs.items():
        print(f"{k} : {v}")
add_kwargs(a=10,b=10,c=5,z=100,book="atomic habit")
```
