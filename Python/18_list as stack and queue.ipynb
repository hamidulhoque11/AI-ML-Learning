```python
value=list(range(1,11))
print(value)
print(type(value))
```

```python
'''
stack : last in last out
'''
print(value)
value.append(11) #last in
print(value)
value.pop() #last out
print(value)
```

```python
#Queue : first in first out
print(value)
value.append(11)#add 11 in last
print(value)
value.pop(0) #remove first index 
print(value)
```

```python
#also we can remove any value from list
value.remove(6)# romove 6
print(value)
```
