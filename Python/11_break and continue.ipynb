```python
sum=0
for x in range(1,11):
    
    if x%2==0:
        continue
    print(x)
    sum+=x
print(sum)
```

```python
sum=0
i=1
while i <= 10:
    if i%2 != 0:
        i+=1
        continue
    print(i)
    sum+=i
    i+=1
print("total sum:",sum)
```
