```python
sum=0
for x in range(1,11):
    
    if x%2==0:
        continue
    print(x)
    sum+=x
print(sum)
```

```python
sum=0
i=1
while i <= 10:
    if i%2 != 0:
        i+=1
        continue
    print(i)
    sum+=i
    i+=1
print("total sum:",sum)
```

```python
accuracy=80
for x in (range(1,101)):
    accuracy+=1
    print(accuracy)
    print(x)
    if accuracy==100:
        break
print(accuracy)
```
