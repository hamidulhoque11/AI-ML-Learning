```python
com={x:x**2 for x in range(1,11) if x%2==0}
print(com)
print(type(com))
```

```python
location=[(10,20,30),(40,50,60),(70,80,90)]
print(type(location))
address=["Dhaka","Cumilla","Chattogram"]
print(location,address)
```

```python
#dictonary
addjust={ loc:add for loc,add in zip(location,address) }#ekhane zip er maddo me 2ta itaration ekste korte parchi
print(addjust)
print(type(addjust))
```

```python
print(addjust.get(((10, 20, 30))))
```

```python
id=[1,2,3,4]
name=["hamidul","sohan","apurba","mufa"]
print(type(id))
print(type(name))
id_name={id :name for id, name in zip(id,name)}
print(id_name)
print(type(id_name))
```
