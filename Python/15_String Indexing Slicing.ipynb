Indexing

```python
string="hamidul hoque masum"
length=len(string)
print(length)
for i in range(length):    
    print(string[i],end=" ")
    print(i)
print()
print(string[0])
```

Slicing

```python
massage = """my name is hamidul hoque masum
i am 25 year old, i study in southeast university
and i am gonna be the greatest ai engineer 
"""
print(massage)
first_massage=massage[0:30]
print("first massage:",first_massage)
```
