```python
from functools import reduce
with open("D:/play ground/AI-ML-Learning/Python/31_readFile.txt","r")as file:
    string=file.readlines()
    
    #total line
    total_lines=len(string)
    print(total_lines)
    
    #total word
    total_word_inline=list(map(lambda x : len(x.split(),),string))
    print(total_word_inline)
    
    total_word= reduce(lambda x,y: x+y,total_word_inline)
    print(total_word)
    
    #cleaning extra word
    string =list(map(str.strip,string))
    print(string)
    
    #space remove
    string=list(map(lambda x: x.replace(" ",""),string))
    print(string)
    
    #total charactre
    total_character_in_line=list(map(lambda x:len(x),string))
    print(total_character_in_line)
    
    total_character=reduce(lambda x,y:x+y, total_character_in_line)
    print(total_character)

with open("D:/play ground/AI-ML-Learning/Python/31_wite a new File.txt","w")as file:
    file.write(f"Total No of Line: {total_lines}.\nTotal No of Word: {total_word}.\nTotal No of Character: {total_character}.")
```
