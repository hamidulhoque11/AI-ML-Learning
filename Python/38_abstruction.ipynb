```python
from abc import ABC, abstractmethod

class Telephone(ABC):

    @abstractmethod
    def make_call(self):
        pass


class Sphone(Telephone):
    def make_call(self):
            print("making a call using smartphone")

class Iphone(Telephone):
    def make_call(self):
            print("making a call using iphone")
```

```python
ip=Iphone()
ip.make_call()
```
