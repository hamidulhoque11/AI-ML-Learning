Spliting

```python
prompt="what is phitron ?"#string
token=prompt.split()#convert to list
print(token)
```

```python
prompt="what-is-phitron-?"#string
token=prompt.split("-")#convert to list
print(token)
```

```python
prompt="what,is,phitron-?"#string
token=prompt.split(",")#convert to list
print(token)
```

```python
prompt="what is your name ?"
token=prompt.split()
print(token)
```

Joining

```python
sentence="-".join(token)
print(sentence)
```

```python
print(type(token),type(sentence))
print(token[0])
print(type(token[0]))
```
