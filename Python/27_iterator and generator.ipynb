```python
lst = [x for x in range(500)]

def data_leader(chunk_size, lst):
    for i in range(0, len(lst), chunk_size):
        '''
        loop ta 0 theke star hoiche 
        list er lenth porjonto sesh hoiche 
        prottekbar chunk size 5 kore change hobe
        '''
        yield lst[i:i + chunk_size]# i=0 : i=0+5 tahole 0,1,2,3,4 er ekta list slising hoye return korbe 

x = data_leader(5, lst)

print(next(x))
print(next(x))
print(next(x))
```
