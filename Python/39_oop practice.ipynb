```python
class student_info:
    name=""
    id=""
    cgpa=""
    def display(self):
        print(f"Name : {self.name} | ID : {self.id} | CGPA : {self.cgpa}")

studentOne=student_info()
studentOne.name="hamidul"
studentOne.id=10005
studentOne.cgpa=3.00
#call display method from student info class
studentOne.display()

#2nd student
studentTwo=student_info()
studentTwo.name="Meskat Merad"
studentTwo.id=10006
studentTwo.cgpa=3.10
#call display method
studentTwo.display()
```

constructor

```python
class Student:
    name=""
    id=""
    cgpa=""
    def __init__(self,name,id,cgpa):
        self.name=name
        self.id=id
        self.cgpa=cgpa
    def display(self):
        print(f"Name : {self.name} | ID : {self.id} | Cgpa : {self.cgpa:.2f} .")

studentOne=Student("hamidul hoque masum",10005,3.000)
studentOne.display()

studentTwo=Student("Mufassir Mahmud",10060,3.20)
studentTwo.display()
```

```python
class triangle:
    def __init__(self,base:int,height:int):
        self.base=base
        self.height=height
    def calculate_area(self):
        r= 0.5*self.height*self.base
        return (r)
tri=triangle(10,20)
value=tri.calculate_area()
print(value)
```
