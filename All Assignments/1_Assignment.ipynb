problem 1 : Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

Input Format:
Space-separated integers

Output Format:
A list of numbers satisfying the condition

Example:
Input:
4 6 8 12 16 20 24 28 

Output:
[4, 12, 20, 28] 


```python
num=input().split()
value=[]
for x in num:
    i=int(x)
    if i%4==0 and i%8!=0:
        value.append(i)
print(value)
```

```python
num=input().split()
num=[int(x) for x in num]
div=list(filter(lambda  x : x%4==0 and x%8!=0,num))
print(div  )
```
