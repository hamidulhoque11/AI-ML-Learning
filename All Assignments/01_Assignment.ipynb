problem 1 : Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

Input Format:
Space-separated integers

Output Format:
A list of numbers satisfying the condition

Example:
Input:
4 6 8 12 16 20 24 28 

Output:
[4, 12, 20, 28] 


```python
num=input().split()
value=[]
for x in num:
    i=int(x)
    if i%4==0 and i%8!=0:
        value.append(i)
print(value)
```

```python
num=input().split()
num=[int(x) for x in num]
div=list(filter(lambda  x : x%4==0 and x%8!=0,num))
print(div  )
```

problem 2: Write a function compute_ratio(x, y) that returns (x + y) / (x * y). Use exception handling to manage division by zero.
Use exception handling to manage:
Division by zero

Input Format:
Two values

Output Format:
Computed result OR error message

Example:
Input:
5 0

Output:
Error: Division not possible


```python
x,y=map(int,input().split())
try:
    def compute_ratio(x, y):
        return(x+y)/(x*y)
    compute_ratio(x,y)
except ZeroDivisionError:
    print("Error: Division not possible")
else:
    print(compute_ratio(x,y))
finally:
    print("gpu stop")
```

Problem 3: Write a Python program that reads a text file and finds the longest word.

Requirements:

Open and read data from a file (Example: data.txt)
If multiple words have the same longest length, print any one of them


Input Format:
A text file containing multiple lines of text

Output Format:
The longest word

Example File Content (data.txt):
Python is a wonderful language for machine learning.

Output:
wonderful


```python
with open("D:/hamidul play ground/AI-ML-Learning/All Assignments/01_Assignment data.txt","r")as file:
    data=file.readlines()
    print(data)
    data=list(map(str.split,data))
    print(data)
    longest_len=len(data[0][0])
    longest_word=""
    for x in data:
        for i in x:
            if len(i) > longest_len:
                longest_word=i
                longest_len=len(i)
    print(longest_word)

```

problem 4:Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

Input Format:

A single line sentence

OutputFormat:
A list of strings

Example:
Input:
An apple a day keeps the doctor away
Output:
['An', 'apple', 'a', 'away']


```python
string=input()
print(string)
lst=string.split()
final_lst=[]
for x in lst:
    if x[0]=='a' or x[0]=='e' or x[0]=='i' or x[0]=='o' or x[0]=='u' or x[0]=='A' or x[0]=='E' or x[0]=='I' or x[0]=='O' or x[0]=='U':
        final_lst.append(x)

print(final_lst)
```

 problem 5: A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

Input Format:
A single word string

Output Format:
"Palindrome" or "Not Palindrome"

Example:
Input:
Radar

Output:
Palindrome



```python
string=input()
value=""
for x in range(-1,-(len(string)+1),-1):
    value+=string[x]
if string.lower()==value.lower():
    print("Palindrome")
else:
    print("Not Palindrome")
```
