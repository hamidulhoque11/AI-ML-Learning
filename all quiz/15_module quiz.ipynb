1. Which function is used to draw a simple line graph?

plt.plot()

2. Which function is used for a scatter plot?

plt.scatter()

3. Which function helps to make a bar chart?

plt.bar()

4. Which function is used for showing data distribution?

plt.hist()

5. Which function is used for pie chart?

plt.pie()

6. What does plt.show() do?

Display the plot

7. Which function adds a title to a plot?

plt.title

8. Which functions are used to name the axes?

plt.xlabel() and plt.ylabel()

9. What type of data does a bar chart show?

Categorical data

10. What type of data does a histogram show?

Continuous numerical data
