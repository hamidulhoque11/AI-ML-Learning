#min max scalling

```python
import pandas as pd
df=pd.read_csv('new_student_data.csv')
mn=df['Total Marks'].min()
mx=df['Total Marks'].max()
df['Scaled Marks']=df['Total Marks'].apply(lambda x:(x-mn)/(mx-mn)).round(1)
df
```

custom build function

```python
def greating_system(marks):
    if marks>84:
        return 'A+'
    elif marks>79:
        return 'A'
    elif marks >74:
        return 'A-'
    else:
        return 'F'
df['CGPA']=df['Total Marks'].apply(greating_system)
```

```python
df
```

```python
def marking_system(df):
    a=df['Data Structure Marks']*2
    b=df['Algorithm Marks']*3
    c=df['Python Marks']*3
    return a+b+c
df['Exceptional Marks']=df.apply(marking_system,axis=1)
```
