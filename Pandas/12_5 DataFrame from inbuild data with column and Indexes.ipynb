List to pandas dataframe

```python
import pandas as pd
lst = [["hamidul",5],["mufassir",15],["sohan",30]]
print(lst)
print(type(lst))
df_list=pd.DataFrame(lst,columns=["Name","ID"],index=[1,2,3])
df_list
```

Tuple to pandas DataFrame

```python
tup=(("Hamidul",26),("Meskat",19),("Maleka",28))
print(type(tup))
#tuple to dataframe with columns and index
df_tup=pd.DataFrame(tup,columns=["Name","Age"],index=[1,2,3])
print(type(df_tup))
df_tup
```

Dictionary to pandas dataframe

```python
dic={
    101:'hamidul',
    102:'asif',
    103:'sohan'
}
print(dic.items())
print(type(dic))
df_dic=pd.DataFrame(dic.items(),columns=["Id","name"])
df_dic
```

```python
dic={
    "name":["hamidul","sohan","mufassir"],
    "age":[24,24]
}
print(type(dic))
df_dic=pd.DataFrame(dic)
df_dic
```

```python
dic={
    "name":["hamidul","sohan","mufassir"],
    "age":[24,24,24]
}
print(type(dic))
df_dic=pd.DataFrame(dic)
df_dic
```
