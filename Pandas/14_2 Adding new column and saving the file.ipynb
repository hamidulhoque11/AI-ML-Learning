```python
import pandas as pd
import numpy as np
dataset=pd.read_csv('D:/play ground/AI-ML-Learning/Pandas/student_data.csv')
dataset
```

add county column as feaures

```python
dataset['Country']='Bangladesh'
dataset
```

add new Total Marks

```python
dataset['Total Marks']=round((dataset['Data Structure Marks'] + dataset['Algorithm Marks'] + dataset['Python Marks'])/3)
dataset
```

Add Grade column

```python
dataset['Grade']=np.where(dataset['Total Marks']>79,'A+','A')
dataset
```
